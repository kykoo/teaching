## Civil Engineering 

::: {.incremental}

is the profession focused on the design, construction, operation, and maintenance of the built and natural environment that society relies on.

- Public works: roads, bridges, tunnels, dams, airports, railways, water and wastewater systems
- Main objectives: safety, serviceability, resilience, sustainability, and value for money
- Civil engineers turn science and mathematics into practical systems for communities
:::

## What is infrastructure?

::: {.incremental}
Infrastructure is the network of facilities and systems that allows society and the economy to function.

- Hard infrastructure: transport, energy, water, drainage, communications networks
- Soft infrastructure: institutions and services supporting social and economic activity

:::

## Careers in Water Engineering 


:::: {.columns}
::: {.column width=60%}
- Water planning and hydrology
- Hydraulics and flood systems
- Treatment and utility operations
- Environmental protection and compliance
:::

::: {.column width=40%}
![](https://upload.wikimedia.org/wikipedia/commons/2/2a/A0473_Tenerife%2C_water_treatment_aerial_view.jpg){fig-align="center"}

::::

::: {.notes}

- Water planning and hydrology: 
  - water resources engineer, hydrologist, hydroinformatics engineer
- Hydraulics and flood systems: 
  -hydraulic engineer, flood risk engineer, stormwater/drainage engineer, coastal engineer
- Treatment and utility operations: 
  - water treatment engineer, wastewater treatment engineer
- Environmental protection and compliance: 
  - environmental engineer, pollution/remediation engineer, EIA/compliance consultant
:::

## Careers in Geotechnical Engineering 

:::: {.columns}
::: {.column width=60%}
- Site investigation and ground characterization
- Design of foundations and earth structures
- Underground and infrastructure geotechnics
<!-- - Risk, environment, and resilience
- Specialist and delivery roles -->
:::

::: {.column width=40%}
![](https://upload.wikimedia.org/wikipedia/commons/f/f5/Moosfluh_slope_instability_ESA372872.jpg){fig-align="center"}
:::
::::

::: {.notes}
- Site investigation and ground characterization: ground investigation engineer, geotechnical testing engineer, engineering geologist
- Design of foundations and earth structures: geotechnical design engineer (foundations, retaining walls, slopes), earthworks engineer
- Underground and infrastructure geotechnics: tunnel/underground engineer, geotechnical engineer for transport and utilities
- Risk, environment, and resilience: geohazards/slope stability engineer, geoenvironmental engineer (contaminated land/remediation)
- Specialist and delivery roles: offshore geotechnical engineer, numerical geotechnical analyst, geotechnical project engineer/consultant
:::

## Careers in Transportation Engineering



:::: {.columns}
::: {.column width=60%}
- Transport planning and policy
- Traffic operations and safety
- Highway and pavement engineering
- Rail and transit systems
- Airport and airside infrastructure
:::

::: {.column width=40%}
![](https://upload.wikimedia.org/wikipedia/commons/0/0c/Remote_view_of_the_high_speed_rail_viaduct_near_THSR_Yunlin_Station.jpg){fig-align="center"}
:::
::::   

::: {.notes}
- Transport planning and policy: transport planner, mobility analyst, urban transport consultant
- Traffic operations and safety: traffic engineer, road safety engineer, signal systems engineer
- Highway and pavement engineering: highway design engineer, pavement engineer, drainage engineer
- Rail and transit systems: rail infrastructure engineer, metro/light-rail engineer, station design engineer
- Airport and airside infrastructure: airport civil engineer, runway/taxiway engineer
:::

- Future of Civil Engineering
  - Ageing infrastructure retrofit, steady governmental infrastructure investment
  - digitalisation, sustainability, climate adaptation, automation and intelligent infrastructure

- Proud of Civil Engineers
