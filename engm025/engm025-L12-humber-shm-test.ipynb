## Humber Bridge: Key Specifications


- **Opened** in 1981, located at Hessle (East Yorkshire) to Barton-upon-Humber (North Lincolnshire), UK 
- **Total length:** 2.22 km, longest in the world for 17 years, but still the longest among ones with **pedestrian path**.
- **Main span:** 1,410 m, **Side spans:** 280 m (N), 530 m (S)
- **Tower height:** about 155 m, **Deck width:** 28.5 m
- **Main cables:** 2 cables, diameter 0.68 m; each with 14,948 wires, **Total wire length:** about 71,000 km (44,000 miles)



# History of Monitoring on Humber Bridge

## Structural Monitoring (1990-1991): Humber Bridge 

*University of Bristol* and *Politecnico di Milano* collaboration

- Main purpose: calibrate modelling assumptions for the Messina Straits Bridge project.
- One of the earliest long-span bridge monitoring campaigns using electric signal cables.
- Data were collected manually and stored on 1 GB optical disk drives.
- Field setup included: zoom lens and telescopes, tracked optical targets ("the bird"), wind vector measurements, and extensive signal cabling (about 30 km)
